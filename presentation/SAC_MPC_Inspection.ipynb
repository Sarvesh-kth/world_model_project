# Imagined SAC versus Calle MPC: GIFs, plots and ZIP

Reads existing recordings; no robot runs, training or GPU model loading. Six galleries: two SAC policies and Calle MPC, each on empty and obstacle tables. Each selects up to 4 random successes, 3 closest failures and 3 other random failures, without duplicates. Final plots use all 180 episodes after all six groups finish.

If Matplotlib is missing, run `%pip install matplotlib` in a separate notebook cell first. Rerun setup after pulling updated viewer code.

Pairing checks verify actual saved scene layouts, initial states, planned seeds, and current world-model hashes against SAC provenance. MPC legacy recordings lack code/model hashes captured at runtime; this limit remains in the exported provenance. Controllers have different obstacle inputs/scoring: compare complete systems rather than claim an isolated planning effect.


## 1. Setup — run first

```python
from pathlib import Path
import importlib.util

project = Path.home() / "Robots&EmbodiedAI/project/world_model_sac_obstacles"
module_path = project / "presentation/sac_viewer.py"
spec = importlib.util.spec_from_file_location("sac_viewer", module_path)
module = importlib.util.module_from_spec(spec)
spec.loader.exec_module(module)
viewer = module.SACViewer(
    project / "simulation/data/calle_sac_scratch_v1",
    mpc_runs={
        "empty": project / "simulation/data/calle_mpc_matched_empty30",
        "obstacles": project / "simulation/data/calle_mpc_matched_obstacles30",
    },
)
print("Outputs:", viewer.output)
```
